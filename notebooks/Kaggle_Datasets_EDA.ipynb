```python
RAW = "data/raw"
CLEAN = "data/clean"
os.makedirs(CLEAN, exist_ok=True)
```

```python
FLAG = {1.0: True, 0.0: False}
```

```python
def collect(row, cols):
    return [str(row[c]).strip() for c in cols if pd.notna(row[c])]
```

```python
def split_names(value):
    if pd.isna(value):
        return []
    return [p.strip() for p in str(value).split(",") if p.strip()]

```

```python
def write_jsonl(df, name):
    df.to_json(f"{CLEAN}/{name}.jsonl", orient="records", lines=True, force_ascii=False)
    print(name, len(df))
```

```python
b = pd.read_csv(f"{RAW}/battles.csv")
```

```python
battles_clean = pd.DataFrame({
    "battle_number": b["battle_number"],
    "name": b["name"],
    "year": b["year"],
    "attacker_king": b["attacker_king"],
    "defender_king": b["defender_king"],
    "attackers": b.apply(lambda r: collect(r, [f"attacker_{i}" for i in range(1, 5)]), axis=1),
    "defenders": b.apply(lambda r: collect(r, ["defender_1", "defender_2"]), axis=1),
    "attacker_commanders": b["attacker_commander"].map(split_names),
    "defender_commanders": b["defender_commander"].map(split_names),
    "outcome": b["attacker_outcome"],
    "battle_type": b["battle_type"],
    "major_death": b["major_death"].map(FLAG),
    "major_capture": b["major_capture"].map(FLAG),
    "attacker_size": b["attacker_size"],
    "defender_size": b["defender_size"],
    "summer": b["summer"].map(FLAG),
    "location": b["location"],
    "region": b["region"],
    "note": b["note"],
})
```

```python
write_jsonl(battles_clean, "battles")
```

```python
d = pd.read_csv(f"{RAW}/character-deaths.csv", keep_default_na=False, na_values=[""])
```

```python
d["allegiance"] = d["Allegiances"].replace("None", pd.NA).str.replace(r"^House ", "", regex=True)
```

```python
dup = d["Name"].duplicated(keep=False)
d["name_key"] = d["Name"]
d.loc[dup, "name_key"] = d["Name"] + " (" + d["allegiance"].fillna("None") + ")"
```

```python
flags = ["GoT", "CoK", "SoS", "FfC", "DwD"]
present = d[flags].eq(1)
```

```python
d["first_book"] = (present.values.argmax(axis=1) + 1).astype(float)
d.loc[~present.any(axis=1), "first_book"] = float("nan")
```

```python
chars = pd.DataFrame({
    "name_key": d["name_key"],
    "name": d["Name"],
    "allegiance": d["allegiance"],
    "male": d["Gender"].eq(1),
    "noble": d["Nobility"].eq(1),
    "first_book": d["first_book"].astype("Int64"),
    "intro_chapter": d["Book Intro Chapter"].astype("Int64"),
    "is_dead": d[["Death Year", "Book of Death", "Death Chapter"]].notna().any(axis=1),
    "death_year": d["Death Year"].astype("Int64"),
    "death_book": d["Book of Death"].astype("Int64"),
    "death_chapter": d["Death Chapter"].astype("Int64"),
})
```

```python
write_jsonl(chars, "characters")
```

```python
import json
```

```python
def read_jsonl(path):
    with open(path, encoding="utf-8") as f:
        return [json.loads(line) for line in f if line.strip()]
```

```python
pages = read_jsonl(f"{RAW}/pages.jsonl") + read_jsonl(f"{RAW}/pages_extra.jsonl")
```

```python
by_title = {}
for p in pages:
    by_title[p["title"]] = p
    by_title[p["requested_title"]] = p
```

```python
chars = pd.read_json(f"{CLEAN}/characters.jsonl", lines=True)
chars["match"] = chars["name"].isin(by_title)
```

```python
chars["disambig"] = chars["name"].map(
    lambda n: n in by_title and any("Disambiguation" in c for c in by_title[n]["categories"])
)
```

```python
print(chars["match"].sum(), len(chars), round(chars["match"].mean(), 3))
print(chars["disambig"].sum())
print(chars.groupby("first_book")["match"].mean().round(2))
print(chars[~chars["match"]]["name"].head(40).tolist())
```
